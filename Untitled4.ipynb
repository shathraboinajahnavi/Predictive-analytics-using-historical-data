```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

# ==========================================
# 1. GENERATE MOCK HISTORICAL DATA (For Testing)
# ==========================================
# If you have your own CSV file, you would replace this section with:
# df = pd.read_csv('your_file.csv')

np.random.seed(42)
dates = pd.date_range(start="2024-01-01", periods=100, freq="D")
# Creating a trend: base value growing over time with some random noise/fluctuation
historical_values = 100 + (np.arange(100) * 1.5) + np.random.normal(0, 10, 100)

df = pd.DataFrame({"Date": dates, "Value": historical_values})

print("--- Step 1: Historical Data Loaded ---")
print(df.head(), "\n")

# ==========================================
# 2. CLEAN & PREPROCESS DATA
# ==========================================
# Check for missing values and fill them if they exist
df["Value"] = df["Value"].fillna(df["Value"].mean())

# Models can't read dates directly, so we convert time into a numeric sequence (0, 1, 2...)
df["Time_Index"] = np.arange(len(df))

# ==========================================
# 3. BUILD THE PREDICTIVE MODEL
# ==========================================
# Prepare inputs for the model
X = df[["Time_Index"]]  # Predictor (Time)
y = df["Value"]         # Target (What we want to forecast)

model = LinearRegression()
model.fit(X, y)

# Predict on our historical data to see how well it fits
df["Predicted_Value"] = model.predict(X)

# ==========================================
# 4. FORECAST THE FUTURE TREND
# ==========================================
# Let's predict the next 30 days into the future
future_days = 30
last_index = df["Time_Index"].max()

future_indices = np.arange(last_index + 1, last_index + 1 + future_days).reshape(-1, 1)
future_predictions = model.predict(future_indices)

# Create a dataframe for the future forecast
future_dates = pd.date_range(
    start=df["Date"].max() + pd.Timedelta(days=1), periods=future_days, freq="D"
)
df_future = pd.DataFrame(
    {"Date": future_dates, "Forecast": future_predictions}
)

# ==========================================
# 5. EVALUATE ACCURACY & VISUALIZE
# ==========================================
mae = mean_absolute_error(df["Value"], df["Predicted_Value"])
r2 = r2_score(df["Value"], df["Predicted_Value"])

print("--- Model Evaluation Metrics ---")
print(f"Mean Absolute Error (MAE): {mae:.2f} (Average mistake the model makes)")
print(f"R-squared Score: {r2:.2f} (1.0 is perfect fit, higher is better)\n")

# Plotting the results
plt.figure(figsize=(12, 6))
plt.plot(df["Date"], df["Value"], label="Actual Historical Data", color="blue", alpha=0.6)
plt.plot(df["Date"], df["Predicted_Value"], label="Model Fit (Historical)", color="orange", linestyle="--")
plt.plot(df_future["Date"], df_future["Forecast"], label="30-Day Future Forecast", color="green", linewidth=2.5)

plt.title("Predictive Analytics: Trend Forecasting", fontsize=14)
plt.xlabel("Date", fontsize=12)
plt.ylabel("Value", fontsize=12)
plt.legend()
plt.grid(True, linestyle=":", alpha=0.6)
plt.show()
```
